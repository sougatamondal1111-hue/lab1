```python

# Graph representing cities and roads

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C'],

    'E': ['F'],
    'F': ['E'],

    'G': ['H'],
    'H': ['G']
}


# ------------------------------------------------
# 1. Check whether a route exists using DFS
# ------------------------------------------------

def route_exists(graph, start, destination):
    visited = set()

    def dfs(city):
        # Destination found
        if city == destination:
            return True

        visited.add(city)

        # Visit all neighbouring cities
        for neighbour in graph[city]:
            if neighbour not in visited:
                if dfs(neighbour):
                    return True

        return False

    return dfs(start)


# ------------------------------------------------
# 2. Find all connected components
# ------------------------------------------------

def find_components(graph):
    visited = set()
    components = []

    for city in graph:

        if city not in visited:
            component = []

            def dfs(city):
                visited.add(city)
                component.append(city)

                for neighbour in graph[city]:
                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)
            components.append(component)

    return components


# ------------------------------------------------
# Test route
# ------------------------------------------------

start = 'A'
destination = 'D'

if route_exists(graph, start, destination):
    print("Route exists from", start, "to", destination)
else:
    print("No route exists from", start, "to", destination)


# ------------------------------------------------
# Find connected components
# ------------------------------------------------

components = find_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print("Component", i, ":", component)
```
