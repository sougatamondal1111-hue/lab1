```python

from collections import deque

# Capacities of the jugs
CAP_A = 4
CAP_B = 3


def water_jug_bfs():
    # Initial state
    start = (0, 0)

    # Queue contains (state, path)
    queue = deque()
    queue.append((start, []))

    # Visited states
    visited = set()
    visited.add(start)

    while queue:
        current, path = queue.popleft()
        a, b = current

        # Goal: exactly 2 litres in either jug
        if a == 2 or b == 2:
            return path

        # All possible operations

        # 1. Fill Jug A
        new_state = (CAP_A, b)
        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Fill Jug A"]
            ))

        # 2. Fill Jug B
        new_state = (a, CAP_B)
        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Fill Jug B"]
            ))

        # 3. Empty Jug A
        new_state = (0, b)
        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Empty Jug A"]
            ))

        # 4. Empty Jug B
        new_state = (a, 0)
        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Empty Jug B"]
            ))

        # 5. Pour Jug A -> Jug B
        amount = min(a, CAP_B - b)
        new_state = (a - amount, b + amount)

        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Pour A -> B"]
            ))

        # 6. Pour Jug B -> Jug A
        amount = min(b, CAP_A - a)
        new_state = (a + amount, b - amount)

        if new_state not in visited:
            visited.add(new_state)
            queue.append((
                new_state,
                path + ["Pour B -> A"]
            ))

    return None


# Find the solution
solution = water_jug_bfs()

# Print the solution
print("Minimum sequence of steps:\n")

a = 0
b = 0

print(f"Step 0: ({a}, {b})")

for i, operation in enumerate(solution, start=1):

    if operation == "Fill Jug A":
        a = 4

    elif operation == "Fill Jug B":
        b = 3

    elif operation == "Empty Jug A":
        a = 0

    elif operation == "Empty Jug B":
        b = 0

    elif operation == "Pour A -> B":
        amount = min(a, 3 - b)
        a -= amount
        b += amount

    elif operation == "Pour B -> A":
        amount = min(b, 4 - a)
        b -= amount
        a += amount

    print(f"Step {i}: {operation} -> ({a}, {b})")
```
